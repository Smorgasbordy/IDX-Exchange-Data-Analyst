Submit a .py script that concatenates all monthly files from January 2024 through the most recently completed calendar month into two combined datasets (listings and sold), filters both to PropertyType == ‘Residential’ only, and saves them as new CSVs.
```python
import pandas as pd
```

## Listings

#### Getting Listing Data

```python
# Read in listings: Jan, 2024 - April, 2026

list_24_1 = pd.read_csv("../csv/CRMLSListing202401.csv")
list_24_2 = pd.read_csv("../csv/CRMLSListing202402.csv")
list_24_3 = pd.read_csv("../csv/CRMLSListing202403.csv")
list_24_4 = pd.read_csv("../csv/CRMLSListing202404.csv")
list_24_5 = pd.read_csv("../csv/CRMLSListing202405.csv")
list_24_6 = pd.read_csv("../csv/CRMLSListing202406.csv")
list_24_7 = pd.read_csv("../csv/CRMLSListing202407.csv")
list_24_8 = pd.read_csv("../csv/CRMLSListing202408.csv")
list_24_9 = pd.read_csv("../csv/CRMLSListing202409.csv")
list_24_10 = pd.read_csv("../csv/CRMLSListing202410.csv")
list_24_11 = pd.read_csv("../csv/CRMLSListing202411.csv")
list_24_12 = pd.read_csv("../csv/CRMLSListing202412.csv")
list_25_1 = pd.read_csv("../csv/CRMLSListing202501.csv")
list_25_2 = pd.read_csv("../csv/CRMLSListing202502.csv")
list_25_3 = pd.read_csv("../csv/CRMLSListing202503.csv")
list_25_4 = pd.read_csv("../csv/CRMLSListing202504.csv")
list_25_5 = pd.read_csv("../csv/CRMLSListing202505.csv")
list_25_6 = pd.read_csv("../csv/CRMLSListing202506.csv")
list_25_7 = pd.read_csv("../csv/CRMLSListing202507.csv")
list_25_8 = pd.read_csv("../csv/CRMLSListing202508.csv")
list_25_9 = pd.read_csv("../csv/CRMLSListing202509.csv")
list_25_10 = pd.read_csv("../csv/CRMLSListing202510.csv")
list_25_11 = pd.read_csv("../csv/CRMLSListing202511.csv")
list_25_12 = pd.read_csv("../csv/CRMLSListing202512.csv")
list_26_1 = pd.read_csv("../csv/CRMLSListing202601.csv")
list_26_2 = pd.read_csv("../csv/CRMLSListing202602.csv", low_memory=False)
list_26_3 = pd.read_csv("../csv/CRMLSListing202603.csv")
list_26_4 = pd.read_csv("../csv/CRMLSListing202604.csv")
```

```python
# Create list for easier access

listings_list = [list_24_1, list_24_2, list_24_3, list_24_4, list_24_5, list_24_6,
                 list_24_7, list_24_8, list_24_9, list_24_10, list_24_11, list_24_12,
                 list_25_1, list_25_2, list_25_3, list_25_4, list_25_5, list_25_6,
                 list_25_7, list_25_8, list_25_9, list_25_10, list_25_11, list_25_12, 
                 list_26_1, list_26_2, list_26_3, list_26_4]
```

```python
# Row counts for each month, before concat

year = 2024
month = 1
count = 0
row_counts = []

for i in listings_list:
    rows = i.shape[0]
    row_counts.append(rows)
    
    print(f"Row Count for {month}, {year}: {rows}")
    
    count += 1
    if count == 12:
        count = 0
        year += 1
        month = 0
    month += 1
```

#### Concatenation

```python
# Row count for full concat

listings = pd.concat(listings_list)
print(f"Row Count for all listings: {listings.shape[0]}")
```

```python
# Check that pre- and post-concat rows are the same

print(f"pre/post-concat rows: {sum(row_counts)} vs. {listings.shape[0]}")
```

#### Filtering

```python
# Filtering PropertyType == Residential

filtered_listings = listings[listings["PropertyType"] == "Residential"]
print(f"Row Count for Only Residential Listings: {filtered_listings.shape[0]}")
```

```python
print(f"Row count dropped by {sum(row_counts) - filtered_listings.shape[0]} rows after filtering")
```

```python
# Create csv of filtered data

filtered_listings.to_csv("filtered_listings.csv", index=False)
```
